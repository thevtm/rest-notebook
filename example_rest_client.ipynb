# REST Notebook
Interactive REST client using `requests` and `httpx`. Run cells individually to inspect responses.

```python
import requests
import httpx
from IPython.display import JSON
```

## requests

```python
# GET - fetch a single resource by ID
response = requests.get("https://jsonplaceholder.typicode.com/todos/1")
response.raise_for_status()
JSON(response.json())
```

```python
# POST - json= serializes the payload and sets Content-Type: application/json automatically
payload = {"title": "foo", "body": "bar", "userId": 1}
response = requests.post("https://jsonplaceholder.typicode.com/posts", json=payload)
response.raise_for_status()
JSON(response.json())
```

## httpx

```python
# GET - raise_for_status() raises an exception on 4xx/5xx responses
r = httpx.get("https://jsonplaceholder.typicode.com/todos/1")
r.raise_for_status()
JSON(r.json())
```

```python
# POST - same api as requests, json= handles serialization
payload = {"title": "foo", "body": "bar", "userId": 1}
r = httpx.post("https://jsonplaceholder.typicode.com/posts", json=payload)
r.raise_for_status()
JSON(r.json())
```
